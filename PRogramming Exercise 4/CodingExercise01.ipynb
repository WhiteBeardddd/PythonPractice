```python
# Exercise 1: Analyze a Noisy Signal

import numpy as np
import matplotlib.pyplot as plt

# Step 1: Generate a clean signal (from the example)
Fs = 1000  # Sampling frequency (samples per second)
T = 1/Fs   # Sampling interval
t = np.arange(0, 1, T)  # Time vector (1 second)

# Create a signal with 50 Hz and 120 Hz components
signal = np.sin(2*np.pi*50*t) + np.sin(2*np.pi*120*t)

# Add random noise using np.random.randn()
noisy_signal = signal + 2 * np.random.randn(len(t))

# Step 2: Compute the FFT
n = len(noisy_signal)
f = np.fft.fftfreq(n, T)
fft_values = np.fft.fft(noisy_signal)

# Step 3: Plot results
plt.figure(figsize=(12, 8))

# Plot the noisy signal
plt.subplot(2, 1, 1)
plt.plot(t, noisy_signal, color='blue')
plt.title('Noisy Signal (Time Domain)')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')

# Plot the magnitude of the FFT
plt.subplot(2, 1, 2)
plt.plot(f[:n//2], np.abs(fft_values)[:n//2] * 1/n, color='red')
plt.title('Magnitude Spectrum (Frequency Domain)')
plt.xlabel('Frequency [Hz]')
plt.ylabel('Amplitude')

plt.tight_layout()
plt.show()
```
